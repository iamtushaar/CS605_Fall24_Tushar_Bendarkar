```python
# Node class for each element in the Binary Search Tree
class TreeNode:
    def __init__(self, contact_name, phone_number):
        self.contact_name = contact_name  # Key: Contact_Name
        self.phone_number = phone_number  # Value: Phone number
        self.left = None  # Left child
        self.right = None  # Right child

# Binary Search Tree class
class BinarySearchTree:
    def __init__(self):
        self.root = None  # Root node is initially None

    # Insert a new node into the BST
    def insert(self, contact_name, phone_number):
        new_node = TreeNode(contact_name, phone_number)
        if self.root is None:
            self.root = new_node  # First node becomes the root
        else:
            self._insert_recursive(self.root, new_node)

    # Recursive function for inserting nodes
    def _insert_recursive(self, current_node, new_node):
        if new_node.contact_name < current_node.contact_name:  # Lexicographically smaller goes to the left
            if current_node.left is None:
                current_node.left = new_node
            else:
                self._insert_recursive(current_node.left, new_node)
        else:  # Lexicographically larger goes to the right
            if current_node.right is None:
                current_node.right = new_node
            else:
                self._insert_recursive(current_node.right, new_node)

    # Search for a phone number by name 
    def search(self, contact_name):
        return self._search_recursive(self.root, contact_name)

    # Recursive function for searching nodes
    def _search_recursive(self, current_node, contact_name):
        if current_node is None:  # Return None if name is not present
            return None
        if contact_name == current_node.contact_name:  # Name matches, return the phone number
            return current_node.phone_number
        elif contact_name < current_node.contact_name:  # Search in the left subtree
            return self._search_recursive(current_node.left, contact_name)
        else:  # Search in the right subtree
            return self._search_recursive(current_node.right, contact_name)

    # In-order traversal to print names and phone numbers in sorted order
    def inorder_traversal(self):
        self._inorder_recursive(self.root)

    # Recursive function for in-order traversal
    def _inorder_recursive(self, current_node):
        if current_node is not None:
            self._inorder_recursive(current_node.left)  # Visit left subtree
            print(f"{current_node.contact_name}: {current_node.phone_number}")  # Visit current node
            self._inorder_recursive(current_node.right)  # Visit right subtree

    # Delete a node by name
    def delete(self, contact_name):
        self.root = self._delete_recursive(self.root, contact_name)

    # Recursive function for deleting nodes
    def _delete_recursive(self, current_node, contact_name):
        if current_node is None:
            return current_node
        
        if contact_name < current_node.contact_name:
            current_node.left = self._delete_recursive(current_node.left, contact_name)
        elif contact_name > current_node.contact_name: 
            current_node.right = self._delete_recursive(current_node.right, contact_name)
        else:
            # Node to be deleted found

            # Case 1: Node with no children (leaf node)
            if current_node.left is None and current_node.right is None:
                return None
            # Case 2: Node with one child
            elif current_node.left is None:
                return current_node.right
            elif current_node.right is None:
                return current_node.left
            # Case 3: Node with two children
            else:
                # Find the in-order successor (smallest node in the right subtree)
                successor = self._find_min(current_node.right)
                # Replace current node's value with the successor's value
                current_node.contact_name = successor.contact_name
                current_node.phone_number = successor.phone_number
                # Delete the successor node
                current_node.right = self._delete_recursive(current_node.right, successor.contact_name)

        return current_node

    # function to find the minimum node in a subtree
    def _find_min(self, current_node):
        while current_node.left is not None:
            current_node = current_node.left
        return current_node

# main 
if __name__ == "__main__":
    bst = BinarySearchTree()
    
    # Inserting names and phone numbers
    bst.insert("Tushar", "123-456-7890")
    bst.insert("Amol", "234-567-8901")
    bst.insert("Zakir", "345-678-9012")
    bst.insert("Khushi", "456-789-0123")
    bst.insert("Pushpak", "567-890-1234")

    # Searching for phone numbers
    print("Search for Tushar:", bst.search("Tushar"))
    print("Search for Khushi:", bst.search("Khushi"))
    # Searching for phone number (not in tree)
    print("Search for Rahul :", bst.search("Rahul"))

    # In-order traversal 
    print("\nIn-order Traversal:")
    bst.inorder_traversal()

    # Deleting a leaf node and printing the tree again
    # bst.delete("Pushpak")
    # print("\nAfter Deleting Pushpak:")
    # bst.inorder_traversal()

    # Deleting a node with one child and printing the tree again
    # bst.delete("Khushi") 
    # print("\nAfter Deleting Khushi:")
    # bst.inorder_traversal()

    # Deleting a node with two child and printing the tree again
    # bst.delete("Tushar")
    # print("\nAfter Deleting Tushar:")
    # bst.inorder_traversal()

    # Deleting a node if it is not present
    # bst.delete("Rahul")
    # print("\nAfter Deleting Rahul:")
    # bst.inorder_traversal()
```
